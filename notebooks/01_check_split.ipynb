# Check the official split

Run this notebook from the repository root, or put that root on `sys.path`.
Point `DATA_DIR` at `cleaned_data` after you extract the dataset.
The first 40 samples of each character, ordered by the number in the filename, are the training pool. Later cells only read the file list.

```python
from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
if not (ROOT / "src").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from src.data_pipeline import build_split_index

DATA_DIR = ROOT / "data" / "cleaned_data"
index = build_split_index(
    DATA_DIR,
    train_per_class=40,
    max_classes=5,
    target_samples_per_class=200,
)
print(index.summary())
print("classes:", " ".join(index.class_names))
```

```python
from pathlib import Path

name = index.class_names[0]
train = [Path(path).name for path in index.train_paths if Path(path).parent.name == name]
test = [Path(path).name for path in index.test_paths if Path(path).parent.name == name]
print(name)
print("train", len(train), train[:3], "...", train[-3:])
print("test", len(test), test[:3])
```
